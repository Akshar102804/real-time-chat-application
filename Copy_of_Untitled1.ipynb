```python
#import the library
from nltk.chat.util import Chat, reflections
```

```python
pairs = [
    ['my name is (.*)', ['hi %1']],
    ['(hi|hello|hey|holla|hola)',['hey there', 'hi there', 'hayy']],
    ['(.*) in (.*) is fun', ['%1 in %2 is indeed fun']],
    ['(.*)(location|city) ?', 'Tokyo,Japan'],
    ['(.*) created you ?', ['Akshar112234 did using NLTK']],
    ['how is the weather in (.*)', ['the weather in %1 is amazing like always']],
    ['(.*)help(.*)',['I can help you']],
    ['(.*) your name ?', ['my name is A.K.S.H.A.R']]

]
```

```python
reflections
```

```python
my_dummy_reflections = {
    'go' : 'gone',
    'hello' : 'hey there'
}
```

```python
chat = Chat(pairs, my_dummy_reflections)
#chat._subtitute('go hello')
chat.converse()
```
